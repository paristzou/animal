# Do Larger Animals Live Longer?
A simple freshman Python data project.


## 1. Read the dataset


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('animals.csv')
df
```

## 2. Calculate the mean lifespan


```python
mean_lifespan = df['Lifespan'].mean()
print('Mean lifespan:', mean_lifespan)
```

## 3. Create a scatter plot


```python
plt.scatter(df['Weight'], df['Lifespan'])
plt.xscale('log')
plt.xlabel('Body weight (kg, log scale)')
plt.ylabel('Average lifespan (years)')
plt.title('Body Weight and Lifespan')
plt.show()
```

## 4. Create a bar chart


```python
df_sorted = df.sort_values('Lifespan')
plt.bar(df_sorted['Animal'], df_sorted['Lifespan'])
plt.xticks(rotation=45, ha='right')
plt.ylabel('Average lifespan (years)')
plt.title('Average Lifespan by Animal')
plt.tight_layout()
plt.show()
```
